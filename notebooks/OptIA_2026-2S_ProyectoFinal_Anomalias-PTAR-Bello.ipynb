## Contexto del Proyecto y Problema Operativo

**Organización y Operación:**
* Proyecto **Plataforma de Optimización Energética (POE) Analítica**, desarrollado por GlobalMVM Ingeniería de Software para Aguas Nacionales (Grupo EPM).
* **Foco de estudio:** Estación de bombeo de entrada de la PTAR Bello (sistema crítico de infraestructura hídrica).

**El Problema Operativo:**
* Necesidad de detectar tempranamente anomalías (deriva o fallo) en los transmisores de nivel redundantes (LT-051004A/B) del foso de bombeo.
* **Situación actual:** La detección es reactiva; depende de umbrales fijos (alarmas) o inspecciones manuales periódicas.
* **Impacto del problema:** Una falla no detectada genera lecturas erróneas que provocan decisiones de arranque/parada incorrectas en las bombas. Esto se traduce en consumo energético innecesario y riesgo real de nivel fuera de rango operativo.

<aside class="notes">
Mensaje clave a decir: "El contexto de nuestro proyecto se enmarca en la Plataforma de Optimización Energética para la PTAR Bello. Actualmente, los transmisores de nivel del foso de bombeo dependen de alarmas reactivas. Si un sensor falla y no se detecta de inmediato, las bombas operan mal, gastando energía innecesaria o poniendo en riesgo los niveles operativos de la planta."
</aside>

## Objetivo de la Solución Analítica

Para mitigar el riesgo operativo y el sobrecosto energético, se propuso el siguiente enfoque:

**Enfoque de Modelado:**
* **Tarea de ML:** Clasificación multiclase para evaluar el estado en tres categorías: sano, en deriva o fallo duro.

**Meta de Optimización:**
* Minimizar los falsos negativos, priorizando detectar el "fallo duro".
* **Premisa de Negocio:** Es preferible tener una falsa alarma (deriva clasificada sobre ruido normal) a que el sistema siga confiando en una lectura corrupta.

<aside class="notes">
Mensaje clave a decir: "Por ello, nuestro objetivo analítico es desarrollar un clasificador multiclase que detecte de forma temprana estados de 'deriva' o 'fallo duro'. Para el negocio, es vital minimizar los falsos negativos del fallo duro, ya que operativamente preferimos lidiar con una falsa alarma a seguir tomando decisiones de bombeo con datos corruptos."
</aside>

## Descripción del Dataset: Características

**Origen y Naturaleza de los Datos:**
* Los datos son de **naturaleza sintética** y provienen de un simulador (Generador Automático de Datasets ML, pestaña EOP-05).
* Se dispone de archivos CSV de telemetría y de episodios de anomalías de sensor.

**Características (Features - $X$):**
* **`ltA_medido`, `ltB_medido`:** Nivel medido por cada transmisor; son la entrada directa para calcular el residual.
* **`lt_residual`:** Residual entre transmisores ($ltA - ltB$), siendo la variable central de la clasificación.
* **`residual_roll_mean`:** Media móvil del residual (ventana de 5); ayuda a separar la "deriva" del ruido normal del sensor.
* **`residual_slope`:** Pendiente del residual (ventana de 5); es la señal que más ayuda a diferenciar entre las 3 clases.

<aside class="notes">
Mensaje clave a decir: "Pasando a los datos, trabajamos con un conjunto sintético extraído de un simulador de la operación. Nos centramos en extraer características basadas en el residual o diferencia entre ambos transmisores, calculando también su media móvil y pendiente en el tiempo para capturar el comportamiento dinámico."
</aside>

## Descripción del Dataset: Objetivo y Retos

**Variable Objetivo ($Y$):**
* **`label`:** Variable que representa el estado del transmisor: **sano / deriva / fallo_duro**. *(Nota: Cuando se aplican condiciones de falla y deriva, la falla dura tiene prioridad).*

**Sesgos y Brechas Importantes:**
* **Desbalance de clases:** La clase "sano" domina ampliamente frente a los eventos de deriva y fallo duro, que son raros.
* **Sesgo sintético:** Al no existir una bitácora real de fallas integrada, el modelo requerirá re-entrenamiento con datos reales antes de pasar a producción.

<aside class="notes">
Mensaje clave a decir: "Nuestra meta es clasificar el estado del sensor en 'sano', 'deriva' o 'fallo duro'. Sin embargo, debemos tener presentes dos grandes retos analíticos aquí: primero, un fuerte desbalance de clases, porque naturalmente los fallos son eventos anómalos y poco frecuentes; y segundo, un sesgo sintético que hace estrictamente necesario un reentrenamiento con datos históricos reales de la planta antes de llevar este modelo a producción."
</aside>

## Estrategia de Modelado

**Selección del Algoritmo Base:**
* Se implementó un ensamble de **Random Forest Classifier** por su robustez ante características no lineales y su interpretabilidad de variables.

**Prevención de Fuga de Datos (Data Leakage):**
* Se descartó el particionamiento aleatorio clásico para la división de datos (Train/Test).
* Se implementó un **corte temporal secuencial basado en episodios** de operación.
* **Objetivo:** Evaluar la capacidad real de generalización del modelo sobre eventos futuros que nunca ha visto, simulando el comportamiento en producción.

<aside class="notes">
Mensaje clave a decir: "Para nuestra estrategia de modelado elegimos un ensamble de Random Forest debido a su capacidad para manejar relaciones no lineales y por ser interpretable. Un punto crítico de nuestra metodología fue cómo dividimos los datos: en lugar del clásico particionamiento aleatorio, que podría causar fuga de información, usamos un corte temporal por episodios. Esto nos asegura que el modelo se evalúe sobre eventos futuros reales que nunca ha visto."
</aside>

## Optimización de Hiperparámetros

Para maximizar el rendimiento y controlar el sobreajuste (*overfitting*), se compararon tres metodologías de búsqueda sobre la métrica de exactitud (*Accuracy*):

* **GridSearchCV (Búsqueda en Cuadrícula):** Búsqueda exhaustiva sobre el espacio paramétrico. Es costosa computacionalmente, pero garantiza encontrar el óptimo global dentro de la cuadrícula definida.
* **RandomizedSearchCV (Búsqueda Aleatoria):** Búsqueda estocástica iterativa. Ofrece un excelente balance entre tiempo de procesamiento y velocidad de convergencia.
* **BayesSearchCV (Búsqueda Bayesiana):** Optimización probabilística. Utiliza los resultados anteriores para decidir de forma inteligente qué hiperparámetro evaluar a continuación, logrando gran eficiencia en espacios complejos.

<aside class="notes">
Mensaje clave a decir: "Para ajustar nuestro algoritmo base, comparamos tres técnicas de optimización de hiperparámetros: GridSearchCV, RandomizedSearchCV y BayesSearchCV. Nuestro objetivo era encontrar el mejor balance entre la precisión predictiva y el costo computacional necesario para entrenar el modelo."
</aside>

## Hallazgo Crítico del Negocio

**El costo de los Falsos Negativos:**
Tal como se definió en el objetivo de optimización, el error más costoso operativamente es clasificar un **fallo duro** como un estado "sano".

**Ceguera Analítica Parcial:**
* Los resultados demuestran que el modelo basado **únicamente en el residual matemático** es excelente para detectar derivas graduales.
* Sin embargo, presenta graves debilidades ante colapsos severos del sensor.
* Al estancarse en un **Recall del 16.67%**, el algoritmo está fallando en detectar más del 80% de los eventos críticos donde el transmisor colapsa emitiendo valores numéricamente plausibles.

<aside class="notes">
Mensaje clave a decir: "Sin embargo, de la tabla anterior emerge nuestro hallazgo crítico de negocio: el Recall para la clase 'Fallo Duro' apenas alcanza el 16.67% en todos los escenarios, fallando en detectar más del 80% de los colapsos. Esto nos indica que un modelo alimentado únicamente por el residual matemático sufre de una grave ceguera analítica cuando el sensor falla drásticamente pero sigue enviando un número creíble al sistema."
</aside>

## Análisis del Fenómeno: Ceguera ante Fallo Duro

**Explicación del Hallazgo Analítico:**

* **Deriva (Naranja):** El diferencial crece de forma progresiva y diagonal. El modelo capta perfectamente esta tendencia suave usando la pendiente y medias móviles.
* **Fallo Duro (Rojo - El Problema):** La gráfica demuestra que el fallo duro **sí dispara abrupta y violentamente la diferencia** matemática (picos verticales de hasta 5m).
* **La Causa de la "Ceguera":** La falla predictiva no ocurre porque el evento se confunda con el ruido del sensor. Ocurre porque este salto repentino **rompe la dinámica temporal**. Un cambio tan extremo e instantáneo destruye el patrón progresivo que esperan las variables de tendencia (pendiente/media). Al carecer de una transición suave, el algoritmo no sabe interpretar el patrón y falla en clasificar la anomalía crítica.

<aside class="notes">
Mensaje clave a decir: "Analizando la gráfica, vemos que las derivas crecen progresivamente y el modelo las capta muy bien. En cambio, los fallos duros generan un salto brutal y repentino. ¿Por qué el modelo es ciego ante ellos? Porque este salto es un evento extremo e instantáneo que altera tan drásticamente el cálculo de las variables en ese milisegundo, que el algoritmo no sabe cómo interpretarlo y falla. Esto demuestra el riesgo operativo de depender únicamente de variables de tendencia temporal ante colapsos repentinos del sensor."
</aside>

## Conclusiones Operativas y Analíticas

El desarrollo del modelo predictivo para la PTAR Bello nos deja las siguientes lecciones:

**1. Viabilidad de la Detección Temprana:**
* El uso de Machine Learning para analizar el residual demuestra ser un enfoque válido y superior a la estrategia actual. El modelo captura derivas graduales mucho antes de que se disparen las alarmas reactivas (umbrales fijos).

**2. Limitaciones Estructurales (Ceguera Analítica Parcial):**
