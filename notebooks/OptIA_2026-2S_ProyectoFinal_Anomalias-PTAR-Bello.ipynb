## Contexto del Proyecto y Problema de Negocio

**Organización y Operación:**
* Proyecto **Plataforma de Optimización Energética (POE) Analítica**, desarrollado por GlobalMVM Ingeniería de Software para Aguas Nacionales (Grupo EPM).
* **Foco de estudio:** Estación de bombeo de entrada de la PTAR Bello (sistema crítico de infraestructura hídrica).

**El Problema Operativo:**
* Necesidad de detectar tempranamente anomalías (deriva o fallo) en los transmisores de nivel redundantes (LT-051004A/B) del foso de bombeo.
* **Situación actual:** La detección es reactiva; depende de umbrales fijos (alarmas) o inspecciones manuales periódicas.
* **Impacto del problema:** Una falla no detectada genera lecturas erróneas que provocan decisiones de arranque/parada incorrectas en las bombas. Esto se traduce en consumo energético innecesario y riesgo real de nivel fuera de rango operativo.

**Objetivo de la Solución Analítica:**
* **Tarea de ML:** Clasificación multiclase para evaluar el estado en tres categorías: sano, en deriva o fallo duro.
* **Meta de optimización:** Minimizar los falsos negativos, priorizando detectar el "fallo duro". Es preferible tener una falsa alarma (deriva clasificada sobre ruido normal) a que el sistema siga confiando en una lectura corrupta.

<aside class="notes">
Mensaje clave a decir: "El contexto de nuestro proyecto se enmarca en la Plataforma de Optimización Energética para la PTAR Bello. Actualmente, los transmisores de nivel del foso de bombeo dependen de alarmas reactivas. Si un sensor falla y no se detecta de inmediato, las bombas operan mal, gastando energía innecesaria o poniendo en riesgo los niveles operativos de la planta. Por ello, nuestro objetivo analítico es desarrollar un clasificador multiclase que detecte de forma temprana estados de 'deriva' o 'fallo duro'. Para el negocio, es vital minimizar los falsos negativos del fallo duro, ya que operativamente preferimos lidiar con una falsa alarma a seguir tomando decisiones con datos corruptos."
</aside>

## Descripción del Dataset

**Origen y Naturaleza de los Datos:**
* Los datos son de **naturaleza sintética** y provienen de un simulador (Generador Automático de Datasets ML, pestaña EOP-05).
* Se dispone de archivos CSV de telemetría y de episodios de anomalías de sensor.

**Características (Features - $X$):**
* **`ltA_medido`, `ltB_medido`:** Nivel medido por cada transmisor; son la entrada directa para calcular el residual.
* **`lt_residual`:** Residual entre transmisores ($ltA - ltB$), siendo la variable central de la clasificación.
* **`residual_roll_mean`:** Media móvil del residual (ventana de 5); ayuda a separar la "deriva" del ruido normal del sensor.
* **`residual_slope`:** Pendiente del residual (ventana de 5); es la señal que más ayuda a diferenciar entre las 3 clases.

**Variable Objetivo ($Y$):**
* **`label`:** Variable que representa el estado del transmisor: **sano / deriva / fallo_duro**. *(Nota: Cuando se aplican condiciones de falla y deriva, la falla dura tiene prioridad).*

**Sesgos y Brechas Importantes:**
* **Desbalance de clases:** La clase "sano" domina ampliamente frente a los eventos de deriva y fallo duro, que son raros.
* **Sesgo sintético:** Al no existir una bitácora real de fallas integrada, el modelo requerirá re-entrenamiento con datos reales antes de pasar a producción.

<aside class="notes">
Mensaje clave a decir: "Pasando a los datos, trabajamos con un conjunto sintético extraído de un simulador de la operación. Nos centramos en extraer características basadas en el residual o diferencia entre ambos transmisores, calculando también su media móvil y pendiente en el tiempo para capturar el comportamiento dinámico. Nuestra meta es clasificar el estado en 'sano', 'deriva' o 'fallo duro'. </aside>

## Estrategia de Modelado y Optimización

**1. Selección del Algoritmo Base:**
* Se implementó un ensamble de **Random Forest Classifier** por su robustez ante características no lineales y su interpretabilidad de variables.
* **Prevención de Data Leakage:** Se descartó el muestreo aleatorio clásico. La división de datos (Train/Test) se realizó mediante un **corte temporal secuencial basado en episodios** de operación, evaluando la generalización sobre eventos futuros que el modelo nunca ha visto.

**2. Técnicas de Optimización de Hiperparámetros Evaluadas:**
Para maximizar el rendimiento y controlar el sobreajuste (overfitting), se compararon tres metodologías sobre la métrica de *Accuracy*:
* **GridSearchCV:** Búsqueda exhaustiva sobre el espacio paramétrico (costosa computacionalmente, pero garantiza encontrar el óptimo global dentro de la cuadrícula).
* **RandomizedSearchCV:** Búsqueda estocástica iterativa (excelente balance entre tiempo de procesamiento y convergencia).
* **BayesSearchCV:** Optimización probabilística que utiliza los resultados anteriores para decidir cuál hiperparámetro evaluar a continuación, logrando eficiencia en espacios de búsqueda complejos.

<aside class="notes">
Mensaje clave a decir: "Para nuestra estrategia de modelado elegimos un Random Forest por su capacidad para manejar relaciones no lineales y por ser interpretable. Un punto crítico de nuestra metodología fue cómo dividimos los datos: en lugar del clásico particionamiento aleatorio, usamos un corte temporal por episodios para evitar la fuga de información (data leakage) y asegurar que el modelo se enfrente a eventos futuros reales. Para ajustar el algoritmo, comparamos tres técnicas de optimización de hiperparámetros: GridSearchCV, RandomizedSearchCV y BayesSearchCV, buscando el mejor balance entre precisión predictiva y costo computacional."
</aside>

## Análisis del Fenómeno: Ceguera ante Fallo Duro

<aside class="notes">

Pero en un fallo duro, en rojo, el sensor colapsa reportando un valor que aún es numéricamente posible. Al no disparar un 'NaN' o una diferencia matemática abrupta, el modelo que depende solo del residual es incapaz de discriminar este fallo del ruido normal del pozo, explicando así nuestra ceguera analítica."
</aside>

## Conclusiones y Próximos Pasos

El desarrollo del modelo predictivo para la PTAR Bello nos deja las siguientes lecciones operativas y analíticas:

**1. Viabilidad de la Detección Temprana:**
* El uso de Machine Learning para analizar el residual entre los transmisores demuestra ser un enfoque válido para clasificar estados (sano, deriva, fallo duro). El modelo captura derivas graduales mucho antes de que se disparen las alarmas reactivas (umbrales fijos).

**2. Limitaciones Estructurales (Ceguera Analítica):**
* Se demostró matemáticamente la **"ceguera analítica ante fallos duros"**. Un enfoque que dependa *únicamente* de la magnitud del residual matemático es incapaz de detectar fallas severas donde el transmisor colapsa pero sigue emitiendo valores plausibles sin arrojar un dato nulo (NaN).

**3. Recomendación de Arquitectura de Datos:**
* Para resolver este vacío, es imperativo subsanar la carencia de **métricas nativas de calidad o disponibilidad** en la telemetría actual. Se requiere integrar al pipeline una señal explícita de "salud" (health status) del dispositivo desde el sistema de control distribuido (DCS).

**4. Ruta hacia Producción:**
* **Re-entrenamiento requerido:** Dada la naturaleza sintética de los datos utilizados (provenientes del simulador), el modelo actualmente no está certificado para producción. 
* **Próximo paso:** Se debe ejecutar un re-entrenamiento validado con datos reales provenientes del historiador de la planta antes de su despliegue definitivo.

<aside class="notes">

Segundo, identificamos una limitación estructural crítica: la ceguera analítica ante fallos duros si solo usamos diferencias matemáticas. Por ello, nuestra recomendación principal a nivel de arquitectura de datos es integrar señales explícitas de 'salud' o disponibilidad del sensor directamente desde el DCS de la planta. Finalmente, la ruta hacia producción exige un reentrenamiento del algoritmo con datos históricos reales, dado el origen sintético de nuestro set actual. Muchas gracias por su atención."
